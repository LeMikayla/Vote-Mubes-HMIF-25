```python
import pandas as pd
import random
import string

def generate_password(length=4):
    """Menghasilkan password acak 4 karakter (huruf kecil & angka)"""
    characters = string.ascii_lowercase + string.digits
    return ''.join(random.choice(characters) for _ in range(length))

def process_mahasiswa_excel(input_filename, output_filename):
    try:
        # 1. Membaca file Excel
        # Pastikan file memiliki kolom 'Nama' dan 'NPM'
        df = pd.read_excel(input_filename)
        
        # 2. Logika Pembuatan Username (Nama Akhir)
        def get_last_name(name):
            if pd.isna(name): return "user"
            parts = str(name).strip().split()
            # Ambil kata terakhir dan ubah ke huruf kecil
            return parts[-1].lower() if parts else "user"

        # 3. Terapkan fungsi ke dataframe
        df['Username'] = df['Nama'].apply(get_last_name)
        df['Password'] = [generate_password(4) for _ in range(len(df))]
        
        # 4. Pilih kolom yang diinginkan saja untuk file output
        output_df = df[['NPM', 'Username', 'Password']]
        
        # 5. Simpan ke CSV
        output_df.to_csv(output_filename, index=False)
        print(f"Berhasil! File '{output_filename}' telah dibuat.")
        
    except Exception as e:
        print(f"Terjadi kesalahan: {e}")

# --- Cara Penggunaan ---
# Ganti 'data_mhs.xlsx' dengan nama file Excel kamu yang asli
process_mahasiswa_excel('data_mhs.xlsx', 'hasil_kredensial.csv')
```

```python
import pandas as pd
import random
import string
import os

def generate_password(length=4):
    characters = string.ascii_lowercase + string.digits
    return ''.join(random.choice(characters) for _ in range(length))

def process_excel_files():
    # Mencari semua file xlsx di folder saat ini
    files = [f for f in os.listdir('.') if f.endswith('.xlsx') and not f.startswith('output_')]
    
    if not files:
        print("Tidak ada file .xlsx yang ditemukan.")
        return

    for file in files:
        print(f"Sedang memproses: {file}...")
        
        # Membaca file excel
        df = pd.read_excel(file)
        
        # Pastikan kolom Nama dan NPM ada (case-insensitive check)
        df.columns = [c.strip().lower() for c in df.columns]
        
        if 'nama' not in df.columns or 'npm' not in df.columns:
            print(f"Skip {file}: Kolom 'nama' atau 'npm' tidak ditemukan.")
            continue

        # 1. Generate Username (Nama kata terakhir)
        # Kita ambil split, ambil index terakhir [-1], dan jadikan lowercase
        df['username'] = df['nama'].apply(lambda x: str(x).split()[-1].lower() if pd.notnull(x) else "user")
        
        # 2. Generate Password
        df['password'] = [generate_password() for _ in range(len(df))]
        
        # 3. Pilih kolom yang diinginkan saja untuk output
        output_df = df[['npm', 'username', 'password']]
        
        # Simpan ke CSV baru
        output_filename = f"output_{file.replace('.xlsx', '.csv')}"
        output_df.to_csv(output_filename, index=False)
        print(f"Selesai! Hasil disimpan di: {output_filename}")

if __name__ == "__main__":
    process_excel_files()
```

```python
import pandas as pd
import random
import string
import os

def generate_password(length=4):
    characters = string.ascii_lowercase + string.digits
    return ''.join(random.choice(characters) for _ in range(length))

def process_excel_files(folder_path, output_filename="semua_mahasiswa_kredensial.csv"):
    all_data = []
    
    # Mencari semua file .xlsx di folder yang ditentukan
    for file in os.listdir(folder_path):
        if file.endswith(".xlsx"):
            file_path = os.path.join(folder_path, file)
            print(f"Memproses file: {file}...")
            
            # Membaca file excel
            df = pd.read_excel(file_path)
            
            # Pastikan kolom 'nama' dan 'npm' ada (case-insensitive)
            df.columns = [col.lower() for col in df.columns]
            
            if 'nama' in df.columns and 'npm' in df.columns:
                for _, row in df.iterrows():
                    nama_lengkap = str(row['nama']).strip()
                    npm = str(row['npm']).strip()
                    
                    # Logika Username: Ambil kata terakhir dari nama, lowercase
                    nama_parts = nama_lengkap.split()
                    username = nama_parts[-1].lower() if nama_parts else "user"
                    
                    # Logika Password: 4 karakter acak
                    password = generate_password(4)
                    
                    all_data.append({
                        "NPM": npm,
                        "Username": username,
                        "Password": password
                    })
            else:
                print(f"Peringatan: Kolom 'nama' atau 'npm' tidak ditemukan di {file}")

    # Menggabungkan semua data dan simpan ke satu CSV
    if all_data:
        final_df = pd.DataFrame(all_data)
        final_df.to_csv(output_filename, index=False)
        print(f"\nSelesai! {len(all_data)} data mahasiswa digabungkan ke '{output_filename}'")
    else:
        print("Tidak ada data yang berhasil diproses.")

# --- Cara Penggunaan ---
# Letakkan semua file .xlsx kamu di dalam satu folder (misal folder 'data_mhs')
# Kemudian jalankan fungsi di bawah ini:
# process_excel_files("jalur/ke/folder/kamu") 
process_excel_files("list pemilih") # Ganti "./" dengan folder tempat file excel berada
```
